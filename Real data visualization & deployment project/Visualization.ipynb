# **Importing Libraries**

```python
import pandas as pd
import plotly.express as px
import calendar
```

# **Inserting/Identifying Data**

```python
df = pd.read_csv('Netflix_cleaned_dataset.csv')
```

### 1. Movies and TV shows counts

```python
type_counting = df.type.value_counts().reset_index()
type_counting
```

```python
movie_counts = df.type[df.type == 'Movie'].count()
tvshows_counts = df.type[df.type == 'TV Show'].count()

print(f'The ratio between movie counts and tv shows counts is : {(movie_counts/tvshows_counts).round(decimals=2)}')
```

```python
px.pie(type_counting,title='Movies and TV Shows added number :'
       ,names='type'
       ,values='count'
       ,template='ggplot2'
       ,hole=0)
```

### 2. Shows added in each year

```python
year_counts = df.year_added.value_counts().reset_index().sort_values(by='year_added')
year_counts
```

```python
px.line(year_counts,x='year_added',y='count'
        ,title='The years adding counts'
        ,labels={'year_added':'Year'}
        ,template='plotly_dark'
        ,markers=True )
```

### 3. Shows added in each month

```python
# This one was with pleasure from MR.Chatgpt🤣
# But truely I get it 
month_order = list(calendar.month_name)[1:]
df['month_name'] = pd.Categorical(df['month_name'], categories=month_order, ordered=True)
```

```python
month_repetitions = df.month_name.value_counts().reset_index().sort_values(by='month_name')
month_repetitions
```

```python
px.scatter(month_repetitions,x='month_name',y='count'
           ,title='The distribution of shows added in each month'
           ,template='simple_white'
           ,labels={'month_name':'Month name' , 'count' : 'counts'})
```

### 4. Shows added in each country

```python
df.dropna(subset='country' , inplace=True)
```

```python
df.country = df.country.apply(lambda x : x.split(', '))
```

```python
countries = []
for country in df.country:
        countries += country
```

```python
top10_countries = pd.Series(countries).value_counts().reset_index().sort_values(by='count',ascending=False).head(10)
top10_countries
```

```python
px.bar(top10_countries,x='index',y='count'
       ,title='Top 10 Countries having shows added :'
       ,labels={'index':'Country','count':'Repetitions'}
       ,template='plotly_dark'
       ,color_continuous_scale='Emrld'
       ,color='count'
       )
```

### 5. Age rating frequencies

```python
df = df[df['age_rating'] != 'Nan']
```

```python
age_ratings = df.age_rating.value_counts().reset_index()
age_ratings
```

```python
px.histogram(age_ratings,x='age_rating',y='count'
             ,template='simple_white'
             ,title='Age rating distribution :'
             ,labels={'sum of count':'counts' , 'age_rating' : 'Age rating'})
```

### 6.Movies duration distribution

```python
movies_durations =  df[df['duration_unit'].str.contains('Minute', na=False)].duration_value.value_counts().reset_index().sort_values(by='count')
movies_durations
```

```python
df.duration_value.max()
```

```python
px.box(movies_durations , x = 'duration_value'
       , title = 'Movies durations distribution : '
       ,template='simple_white'
       ,labels={'duration_value' : 'Movie duration'})
```

### 7.TV shows duration distribution

```python
shows_durations =  df[df['duration_unit'].str.contains('Season', na=False)].duration_value.value_counts().reset_index().sort_values(by='count')
shows_durations
```

```python
px.bar(shows_durations,x='duration_value' , y='count'
       ,template='plotly_dark'
       ,labels={'duration_value' : 'Show duration' , 'count' : 'Counts'}
       ,color='count'
       ,color_continuous_scale='Tealgrn')
```
