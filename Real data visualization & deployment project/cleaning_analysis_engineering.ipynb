# **Feature Engineering**

```python
# We should have a refrence to look at ⬇️
df.head()
```

### Dates

```python
df['year_added'] = df['date_added'].dt.year
```

```python
df['month_name'] = df['date_added'].dt.month_name()
```

```python
df['day_name'] = df['date_added'].dt.day_name()
```

```python
df.drop('date_added',inplace=True,axis=1)
```

### Age rating

```python
df['rating'].value_counts()
```

```python
df.rating = df.rating.apply(lambda x : 'Mature Audience' if x == 'TV-MA' else 'above 14' if x == 'TV-14' else 'Parental guidance suggested' if x == 'TV-PG' else 'Restricted' if x == 'R' else 'above 13' if x == 'PG-13' else 'above 7' if x == 'TV-Y7' else 'All children' if x == 'TV-Y ' else 'Parental guidance suggested' if x == 'PG' else 'General audience' if x == 'TV-G' else 'Nan' if x == 'NR' else 'General audience' if x == 'G' else 'Fantasy violence' if x == 'TV-Y7-FV' else 'above 17' if x == 'NC-17' else 'Unrated' if x == 'UR' else 'Nan' if x == '74 min' or '84 min' or '66 min' else 'Nan')
```

```python
df.rename(columns={'rating':'age_rating'} , inplace=True)
```

### Duration

```python
df.head()
```

```python
# Such a great line BTW !
df[['duration_value', 'duration_unit']] = df['duration'].str.split(' ', expand=True)
```

```python
df.drop('duration',inplace=True,axis=1)
```
